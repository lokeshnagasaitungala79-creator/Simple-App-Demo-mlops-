```python
from src.get_data import read_params
import argparse
import mlflow
from mlflow.tracking import MlflowClient


def log_production_model(config_path):
    config = read_params(config_path)

    # Access dictionary correctly
    mlflow_config = config["mlflow_config"]

    model_name = mlflow_config["registered_model_name"]
    remote_server_url = mlflow_config["remote_server_url"]

    # Set MLflow Model Registry URI
    mlflow.set_registry_uri(remote_server_url)

    # Get experiment runs
    runs = mlflow.search_runs(experiment_ids=["1"])

    if runs.empty:
        print("No MLflow runs found.")
        return

    # Find run with lowest MAE
    lowest = runs["metrics.mae"].min()

    lowest_run_id = runs.loc[
        runs["metrics.mae"] == lowest, "run_id"
    ].iloc[0]

    print("Best MAE:", lowest)
    print("Best Run ID:", lowest_run_id)

    # MLflow client
    client = MlflowClient()

    # Find registered model versions
    model_versions = client.search_model_versions(
        f"name='{model_name}'"
    )

    for mv in model_versions:

        print("\nModel Version:")
        print(mv)

        current_version = mv.version

        if mv.run_id == lowest_run_id:

            print(
                f"Version {current_version} belongs to the best run."
            )

            client.transition_model_version_stage(
                name=model_name,
                version=current_version,
                stage="Production"
            )

            print(
                f"Version {current_version} moved to Production."
            )

        else:

            client.transition_model_version_stage(
                name=model_name,
                version=current_version,
                stage="Staging"
            )

            print(
                f"Version {current_version} moved to Staging."
            )


if __name__ == "__main__":

    args = argparse.ArgumentParser()

    args.add_argument(
        "--config",
        default="params.yaml"
    )

    parsed_args = args.parse_args()

    log_production_model(
        config_path=parsed_args.config
    )
```
